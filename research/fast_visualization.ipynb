## 배송지연에 관련된 변수들 위주로 시각화 그래프 분석결과

### 강수


- 서울지역은 계속 0이다가 28일쯤 스파이크가 튀어서 약 0.5mm으로 강수량이 측정됨. 여기에서는 딱히 트랜드, 추세, 등을 찾을 수 없음. 강수를 예측하기가 매우 어려울 것.
- 인천지역은 28일쯤, 0.2가 됬다가 0.1, 0.2 이렇게 세번의 스파이크가 목격됨, 나머지기간은 전부 0
- 대구지역은 28일쯤 두번에 걸쳐 0.5mm의 스파이크가 관측됨, 나머지기간은 전부 0
- 이천은 29일에 1mm의 강우가 측정되고, 조금 뒤 시점에 0.4 mm의 강우가 관측됨, 나머지는 전부 0

- 종합하면, 모든 지역에 걸쳐 대부분은 강우량이 0이고, 어쩌다가 특정요일에 1회 ~ 3회의 스파이크가 관측됨.
이러한 현상이 타겟이 되면 rere 한 현상에 대한 이진분류가 어려울 것. 
타겟변수로 쓰기 적절치 않다는 판단


### 강수타입

- 강수타입
0 : no
1: rain
2: rain & snow mix
3: snow

- 서울
27일, 29일, 30일
0, 1, 2 가 3회 관측되는데
대부분이 0 

- 인천 
27일 2회 28일 1회 1이 관측
나머지는 다 0이 관측

- 대구
28일 1이 2회 관측
나머지는 다 0

- 이천
28일 1관측 1회
29일 3 관측 1회 2관측 1회
나머지는 다 0

- 이렇게 타입도 범주형인데, 0, 1, 2 ,3 중에서 0이 대부분, 1,2가 매우 소수, 3은 1회 이걸 변수로 쓸수 있을지, 쓴다고 해도 0이 너무 많아서 효과가 있을지 의문이긴 하네

### 온도

- 서울
그래프의 분산이 일정하지 않다. 
그러나 일일 계절성은 보인다.
그리고 트랜드는 살짝 보이긴 하는데 기간이 짧아서 잘 모르겠다. 
19일에서 4월 3일로, 관측기간이 짧아서 트랜드는 잘 모르겠음

- 인천
인천도 분산이 달라지는 기간 이 존재한다. 전반적으로 서울보단 분산이 변화가 덜하다.
특히 3월 27일
일일 계절성은 관측되고, 트랜드는 관측이 어렵다.

- 대구
분산이 서울과 인천애 비해 매우 심하다. 매우 불규칙하다. 
일일 계절성은 보이긴 하다.

- 이천
대구 다음으로 분산이 크다. 계절성도 보이나 트랜드는 보이지 않는다.


### 풍속

- 서울
굉장히 분산이 들쭉날쭉하다.
어떤 계절성을 보이는 것 같긴하다.
트랜드는 전혀 보이지 않는다.

- 인천
인천은 완전 분산도 제멋대로고, 특정한 계절성도 보이지 않는다. 트랜드도 보이지 않는다.

- 대구
분산이 제멋대로다, 일정하지 않다. 그리고 계절성도 \딱히 잘 모르겠다. 트랜드도 당연히 잘 모르겠다.

- 이천
그래도 일정한 계절성은 보인다. 근데 분산이 일정하지 않다. 특히 25일에 갑자기 풍속이 강해지는 시간이 있다.
